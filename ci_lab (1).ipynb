```python
from scipy import stats
#training 10 emplyes
data=[42,55,84,62,25,33,36,78,22,65]
# 95% confident intervel
ci=stats.t.interval(confidence=0.95,
                    df=len(data)-1,loc
                    =np.mean(data),scale
                    =stats.sem(data))
print("95% confidence intervel:")
print("lower",ci[0])
print("upper:",ci[1])
```

```python
import numpy as np
from scipy import stats
#training 30 emplyes
data=[42,55,84,62,25,33,36,78,22,65,
      48,82,33,95,65,39,44,75,55,52,
      9,66,38,64,65,88,45,59,49,82]
population_std=3
# 95% confident intervel
ci=stats.norm.interval(
    0.95,
    loc=np.mean(data),
    scale=population_std/np.sqrt(len(data))
)
print ("number of data:",len(data))
print("mean",np.mean(data))

print("95% confidence intervel:")
print("lower",ci[0])
print("upper:",ci[1])
```
